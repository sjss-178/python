```python
# .shift()  .rank()      .rolling()         .cumSum()
#if we want to see the ysterdays units Sold on coffee data
coffee["yestarday_sales"]=coffee["Units Sold"].shift(1)
coffee.fillna(coffee["yestarday_sales"].mean(),inplace=True)
coffee["per_change"]=coffee["Units Sold"]/coffee["yestarday_sales"]
coffee["rank_day"]=coffee["Units Sold"].rank(ascending=False)

print(coffee)
```
